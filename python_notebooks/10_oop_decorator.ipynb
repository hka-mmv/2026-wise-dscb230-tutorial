Ein Dekorator (Decorator) ist ein mächtiges Konzept in Python, das es ermöglicht, das Verhalten einer Funktion oder Methode zu erweitern oder zu verändern, ohne den eigentlichen Quellcode dieser Funktion anzupassen. Es gibt eine Decorator die nur auf Funktionen innerhalb von Klassen angewendet werden sollen. Die wichtigsten werden hier behandelt.

## @staticmethod
Eine statische Funktion in Python ist eine Funktion in einer Klasse, die nicht abhängig von einer Instanz oder Klassendaten sind. Der decorator wird also verwendet, wenn die Funktion logisch zwar zur class gehört, aber keinen Zugriff auf self oder cls (class) braucht  
--> "Utility/Werkzeuge"

```python
class Calc:
    @staticmethod
    def add(a,b): # Die Funktion wird aufgerufen ohne self oder cls verweise.
        return a+b

res = Calc.add(3,7)
print(res)
```

```python
class Person():

    def __init__(self, name, age, job):
        self.name = name
        self.age = age
        self.job = job

    @staticmethod
    def is_adult(age):
        return age >= 18

print(Person.is_adult(29)) # Benötigt keine Objektdaten oder Instanziierung um zu funktionieren
print(Person.is_adult(16))


p1 = Person("Tom", 24, "Student") # Objekt instanz

p1.is_adult(p1.age) # Zugriff mit Objektinformationen
```

## @classmethod
Die classmethod wird verwendet, wenn eine Funktion keinen Zugriff auf die Instanz braucht, aber auf die cls.

```python
class Car:
    total_cars = 0 # Zähler, wie viele Instanzen es aus der Klasse gibt

    def __init__(self, brand, vmax):
        self.brand = brand
        self.vmax = vmax
        Car.total_cars += 1 # Erhöhen des Klassenzählers beim Instanziieren

c1 = Car("BMW", 320)
print(Car.total_cars) # Klassenzähler
c1.total_cars = 100 # Der Klassenzähler in der Instanz wurde verändert --> nicht machen (referenziert Instanz, nicht die Klassen)
print(c1.total_cars)

print(Car.total_cars) # Klassenzähler


c2 = Car("MB", 280)
print(Car.total_cars) # Wir sehen: Es wurden insgesamt 2 Objekte Instanziiert, gezählt wird also nicht von den Objekten, sondern von der Klasse
```

Hier wird ein Alternativer constructor geschrieben, um die Objektinstanziierung zu vereinfachen

```python
class Car:
    total_cars = 0 # Zähler, wie viele Instanzen es aus der Klasse gibt

    def __init__(self, brand, vmax):
        self.brand = brand
        self.vmax = vmax
        Car.total_cars += 1 # Erhöhen des Klassenzählers beim Instanziieren

    @classmethod
    def auto_vmax(cls, brand): # cls -> class, da die Klasse nicht Instanz
        # Anlegen einer "Datenbank" mit den Fahrzeug-vmax
        db = {"bmw": 320,
              "mb": 280}

        vmax = db.get(brand.lower()) # Abfrage aus der "DB"

        if vmax: # Falls vorhanden
            print(f"log: vmax ist: {vmax}")
        else: # Falls nicht vorhanden
            print("log: kein vmax in der DB! default wird verwendet")
            vmax = 200

        return cls(brand=brand, vmax=vmax) # Rückgabe des Konstrukteurs

    def __str__(self):
        return f"{self.brand} ({self.vmax}km/h)"


bmw = Car.auto_vmax("BMW") # Anwendes des neuen Kontrukteurs statt einfacher init
print(bmw)

bmw2 = Car.auto_vmax("bmW")
print(bmw2)

volvo = Car.auto_vmax("VolvO")
print(volvo)

mb = Car.auto_vmax("Mercedes")
print(mb)

mb2 = Car.auto_vmax("mb")
print(mb2)
```

## @abstractmethod

## Abstract Base Class

```python
from abc import ABC, abstractmethod

class Shape(ABC):
    @abstractmethod
    def area(self):
        pass

    @abstractmethod
    def perimeter(self):
        pass

class Circle(Shape):
    def __init__(self, radius):
        self.radius = radius

    def area(self):
        return 3.14159 * self.radius ** 2

    def perimeter(self):
        return 2 * 3.14159 * self.radius

class Square(Shape):
    def __init__(self, height):
        self.height = height
    def area(self):
        return self.height ** 2
    # def perimeter(self):
    #     return self.height * 2

# circle = Shape() # TypeError: Can't instantiate abstract class Shape
circle = Circle(5)
print(circle.area())

square = Square(3)
```
