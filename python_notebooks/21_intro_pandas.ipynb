Pandas ist die wichtigste Python-Bibliothek für die Arbeit mit tabellarischen Daten (ähnlich wie Excel-Tabellen oder SQL-Datenbanken). 

Grundfunktionen von Python sind:
1. Datenstrukturen: `Series` und `DataFrame`
2. Daten inspizieren: Überblick über große Tabellen gewinnen
3. Auswahl & Filterung: Gezielt auf Zeilen und Spalten zugreifen

## Pandas und NumPy importieren

```python
import numpy as np
import pandas as pd

print(f"Verwendete Pandas-Version: {pd.__version__}")
```

## Series und DataFrame

Pandas baut auf zwei zentralen Objekten auf:
* `Series` (1D): Eine einzelne Spalte mit einem Index (vergleichbar mit einer typisierten Liste oder einem eindimensionalen Array).
* `DataFrame` (2D): Eine vollständige Tabelle aus Zeilen und Spalten, wobei jede Spalte eine `Series` ist.

```python
# 1. Eine einfache Series erstellen
umsatz_quartal = pd.Series(
    [12500, 15800, 14200, 19100], index=["Q1", "Q2", "Q3", "Q4"]
)
print("--- Pandas Series ---")
print(umsatz_quartal)
print("\nWert für Q2:", umsatz_quartal["Q2"])
```

```python
# 2. Einen DataFrame aus einem Python-Dictionary erstellen
daten = {
    "Mitarbeiter": ["Anna", "Ben", "Clara", "David", "Elena", "Felix"],
    "Abteilung": ["IT", "Vertrieb", "IT", "Marketing", "Vertrieb", "IT"],
    "Alter": [28, 34, 41, 25, 38, 31],
    "Gehalt": [62000, 58000, 75000, 49000, 64000, np.nan],  # np.nan = fehlender Wert
    "Homeoffice_Tage": [3, 1, 4, 2, 1, 5],
}

df = pd.DataFrame(daten)
df  # "Magic Command": In Jupyter gibt die letzte Zeile einer Zelle die Tabelle formatiert aus --> sollte bereits bekannt sein!
```

## Daten inspizieren und verstehen

Wenn man reale Datensätze (z. B. über `pd.read_csv('datei.csv')`) lädt, verschafft man sich zuerst einen schnellen Überblick über Struktur, Datentypen und statistische Kennzahlen:
* `df.head(n)` / `df.tail(n)`: Zeigt die ersten bzw. letzten $n$ Zeilen.
* `df.info()`: Zeigt Datentypen, Speicherverbrauch und fehlende Werte.
* `df.describe()`: Berechnet statistische Kennzahlen (Mittelwert, Standardabweichung, Quartile) für alle numerischen Spalten.

```python
print("--- Erste 3 Zeilen ---")
display(df.head(3))

print("\n--- Struktur & Datentypen ---")
df.info()

print("\n--- Statistische Zusammenfassung ---")
display(df.describe())
```

## Daten auswählen und filtern (loc, iloc & Bedingungen)

Für den Zugriff auf Daten gibt es in Pandas drei primäre Wege:
1. Spaltenauswahl: Über den Spaltennamen in eckigen Klammern `df['Spalte']` oder `df[['Spalte1', 'Spalte2']]`.
2. `.loc[Zeilenname, Spaltenname]`: Auswahl über Labels/Namen (inklusive logischer Filter).
3. `.iloc[Zeilenposition, Spaltenposition]`: Auswahl streng über den numerischen Index (wie bei Python-Listen, beginnend bei `0`).

```python
# Einzelne Spalte (gibt eine Series zurück) vs. mehrere Spalten (gibt einen DataFrame zurück)
namen = df["Mitarbeiter"]
teil_tabelle = df[["Mitarbeiter", "Gehalt"]]

# Auswahl mit .iloc (Zeile 0 bis 2, Spalte 0 bis 2)
print("--- Auswahl per Position (.iloc) ---")
display(df.iloc[0:3, 0:2])

# Auswahl mit .loc (Zeile mit Index 2, Spalten 'Mitarbeiter' und 'Abteilung')
print("--- Auswahl per Label (.loc) ---")
print(df.loc[2, ["Mitarbeiter", "Abteilung"]])
```

```python
# Logisches Filtern (Boolean Indexing)
# Finde alle Mitarbeiter aus der IT, die älter als 30 sind
it_erfahren = df[(df["Abteilung"] == "IT") & (df["Alter"] > 30)]

print("--- Gefilterte Daten (IT & Alter > 30) ---")
display(it_erfahren)
```

## Daten bereinigen und transformieren

In der Praxis sind Daten selten perfekt. Oft fehlen Werte (`NaN` = *Not a Number*) oder wir müssen aus bestehenden Spalten neue Kennzahlen berechnen.
* `df.isna().sum()`: Zählt fehlende Werte pro Spalte.
* `df.fillna(wert)`: Ersetzt fehlende Werte (z. B. durch den Mittelwert oder `0`).
* `df.dropna()`: Entfernt Zeilen mit fehlenden Werten komplett.

```python
# 1. Fehlende Werte prüfen
print("Fehlende Werte pro Spalte:\n", df.isna().sum())

# 2. Fehlende Gehälter durch das Durchschnittsgehalt auffüllen
durchschnitts_gehalt = df["Gehalt"].mean()
df["Gehalt"] = df["Gehalt"].fillna(durchschnitts_gehalt)

# 3. Neue Spalte berechnen (Monatsgehalt & Bonus-Kategorie)
df["Monatsgehalt_Brutto"] = (df["Gehalt"] / 12).round(2)
df["Viel_Homeoffice"] = df["Homeoffice_Tage"] >= 3

display(df)
```

## Gruppieren, Aggregieren und Sortieren (groupby)

Mit `groupby()` lassen sich Daten nach Kategorien zusammenfassen (das Split-Apply-Combine-Prinzip, bekannt aus SQL oder Excel-Pivot-Tabellen). Zudem können Tabellen mit `sort_values()` beliebig sortiert werden.

```python
# Durchschnittsgehalt und Durchschnittsalter pro Abteilung berechnen
abteilungs_statistik = (
    df.groupby("Abteilung")[["Gehalt", "Alter"]].mean().round(1)
)
print("--- Durchschnittswerte pro Abteilung ---")
display(abteilungs_statistik)

# Mehrere Aggregationen gleichzeitig mit .agg()
detaillierte_übersicht = df.groupby("Abteilung").agg(
    Anzahl_Mitarbeiter=("Mitarbeiter", "count"),
    Max_Gehalt=("Gehalt", "max"),
    Schnitt_Homeoffice=("Homeoffice_Tage", "mean"),
)
print("\n--- Detaillierte Abteilungs-Aggregation ---")
display(detaillierte_übersicht)

# Tabelle absteigend nach Gehalt sortieren
df_sortiert = df.sort_values(by="Gehalt", ascending=False)
print("\n--- Sortiert nach Gehalt (absteigend) ---")
display(df_sortiert)
```
